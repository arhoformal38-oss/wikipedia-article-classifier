# Live Demo — Wikipedia Article-Type Classifier

Classifies a short English article text as **Company, Artist, Athlete, Building, Film or WrittenWork**
using the Bag of Words + Logistic Regression model from Project 1.

**How to run in Google Colab:** `Runtime → Run all`, then type texts in the box that appears under the last cell.

```python
# 1) Get the project code, data and saved model (only needed in Colab)
import os
if not os.path.exists('demo.py'):
    if not os.path.exists('wikipedia-article-classifier'):
        !git clone -q https://github.com/arhoformal38-oss/wikipedia-article-classifier.git
    %cd wikipedia-article-classifier
!pip -q install nltk scikit-learn pandas joblib
```

```python
# 2) Load the trained model (uses the same preprocessing as training)
import demo
model = demo.load_model()
print('Classes:', list(model.classes_))
```

```python
# 3) Examples written for the demo (not from the corpus)
examples = [
    'Nova Robotics is a technology firm headquartered in Austin Texas that develops warehouse automation software.',
    'Daniel Okafor is a Nigerian professional footballer who plays as a striker in the English Premier League.',
    'The Harbour Tower is a 40-storey skyscraper completed in 1998 and designed by a local architect.',
    'Silent Orbit is a 2019 science fiction thriller directed by James Reed.',
    'The Glass Garden is a novel by a British author published in 2004.',
    'Jonathan Rosenbaum is an American film critic.',   # a hard case: a person, but full of film words
]
for t in examples:
    ranked = demo.predict_article_type(model, t)
    print(f'{ranked[0][0]:<12} {ranked[0][1]:6.1%}   {t}')
```

```python
# 4) Try your own text (type 'q' or leave empty to stop)
while True:
    text = input('Text> ').strip()
    if not text or text.lower() == 'q':
        break
    ranked = demo.predict_article_type(model, text)
    print(f'  Prediction: {ranked[0][0]} ({ranked[0][1]:.1%})   Top 3: ' +
          ', '.join(f'{c} {p:.1%}' for c, p in ranked[:3]))
```
